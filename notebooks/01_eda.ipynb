# Titanic 存活预测 —— 探索性数据分析（EDA）

本 notebook 走通完整流程：数据加载 → 缺失值检查 → 特征与存活率关系可视化 → 特征工程。

运行前提：先把 `train.csv` / `test.csv` 放入 `data/` 目录。

```python
import sys
sys.path.append('../src')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from data_utils import load_data, summarize_missing, impute_age, build_features

sns.set_style('whitegrid')
%matplotlib inline
```

## 1. 加载数据

```python
train, test = load_data()
print('训练集形状:', train.shape)
print('测试集形状:', test.shape)
train.head()
```

## 2. 缺失值检查

```python
summarize_missing(train)
```

## 3. 关键特征与存活率的关系

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 9))
sns.barplot(x='Sex', y='Survived', data=train, ax=axes[0, 0])
axes[0, 0].set_title('性别与存活率')
sns.barplot(x='Pclass', y='Survived', data=train, ax=axes[0, 1])
axes[0, 1].set_title('舱位等级与存活率')
sns.histplot(train['Age'].dropna(), bins=30, ax=axes[1, 0])
axes[1, 0].set_title('年龄分布')
sns.barplot(x='Embarked', y='Survived', data=train, ax=axes[1, 1])
axes[1, 1].set_title('登船港口与存活率')
plt.tight_layout()
plt.show()
```

## 4. 特征工程

构造家庭规模、是否独行、称谓三个新特征，并填补年龄缺失值。

```python
train = impute_age(train)
train = build_features(train)
print('新增列:', [c for c in train.columns if c in ('FamilySize', 'IsAlone', 'Title')])
train[['FamilySize', 'IsAlone', 'Title']].head()
```
